```python
import cv2
import numpy as np

img = cv2.imread('lake.jpg')
cv2.imshow('Input Image', img)
cv2.waitKey(0)
```

```python
gray_img = cv2.imread('lake.jpg', cv2.IMREAD_GRAYSCALE)
cv2.imshow('Grayscale Image', gray_img)
cv2.waitKey(0)
```

```python
cv2.imwrite('lake.jpg', gray_img)
```

```python
gray_from_color = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imshow('BGR to Gray', gray_from_color)
cv2.waitKey(0)
```

```python
yuv_img = cv2.cvtColor(img, cv2.COLOR_BGR2YUV)

cv2.imshow('Y channel', yuv_img[:, :, 0])
cv2.imshow('U channel', yuv_img[:, :, 1])
cv2.imshow('V channel', yuv_img[:, :, 2])

cv2.waitKey(0)
```

```python
#Task 1.1 (Scaling)
scaled_img = cv2.resize(img, None, fx=1.5, fy=1.5)
cv2.imshow('Scaled Image', scaled_img)
cv2.waitKey(0)
```

```python
#Task 1.2 (Rotate Image by 120 degrees)
h, w = img.shape[:2]
center = (w // 2, h // 2)

rotation_matrix = cv2.getRotationMatrix2D(center, 120, 1)
rotated_img = cv2.warpAffine(img, rotation_matrix, (w, h))

cv2.imshow('Rotated Image', rotated_img)
cv2.waitKey(0)
```

```python
#Task 1.3 (Sheer operation)
shear_matrix = np.float32([
    [1, 0.3, 0],
    [0.3, 1, 0]
])

sheared_img = cv2.warpAffine(img, shear_matrix, (w, h))
cv2.imshow('Sheared Image', sheared_img)
cv2.waitKey(0)
```

```python
#Task 2.1 (Negative)
negative_img = 255 - gray_img
cv2.imshow('Negative Image', negative_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#Task 2.2 (Log)
import numpy as np
import matplotlib.pyplot as plt

gray_float = gray_img.astype(np.float32)

c = 255 / np.log(1 + np.max(gray_float))
log_img = c * np.log(1 + gray_float)

log_img = np.uint8(log_img)

plt.imshow(log_img, cmap='gray')
plt.title('Log Transformation')
plt.axis('off')
plt.show()
```
