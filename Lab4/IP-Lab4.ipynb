```python
import matplotlib.pyplot as plt
import numpy as np
from skimage import data
from skimage import exposure
```

```python
moon = data.moon()   # Load moon image
plt.imshow(moon, cmap='gray')
plt.title("Original Moon Image")
plt.axis('off')
plt.show()
```

```python
#Task 1

# Computes percentiles
p3, p80 = np.percentile(moon, (3, 80))

# rescale intensity
moon_rescale = exposure.rescale_intensity(moon, in_range=(p3, p80))

# Display result
plt.figure(figsize=(10,7))

plt.subplot(2,1,1)
plt.imshow(moon_rescale, cmap='gray')
plt.title("Rescaled (3rd–80th Percentile)")
plt.axis('off')

plt.subplot(2,1,2)
plt.hist(moon_rescale.flat, bins=256, range=(0,255))
plt.xlabel("Gray level values")
plt.ylabel("Number of pixels")

plt.tight_layout()
plt.show()
```

```python
#Task 2

# Apply histogram equalization
moon_equalized = exposure.equalize_hist(moon)

# Display result
plt.figure(figsize=(10,7))

plt.subplot(2,1,1)
plt.imshow(moon_equalized, cmap='gray')
plt.title("Histogram Equalized")
plt.axis('off')

plt.subplot(2,1,2)
plt.hist(moon_equalized.flat, bins=256)
plt.xlabel("Gray level values")
plt.ylabel("Number of pixels")

plt.tight_layout()
plt.show()
```

```python
#Task 3

# Load images
reference = data.rocket()
source = data.chelsea()

# Match histogram
matched = exposure.match_histograms(source, reference, channel_axis=-1)

# Display results
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(source)
plt.title("Source (Chelsea)")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(reference)
plt.title("Reference (Rocket)")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(matched)
plt.title("Histogram Matched Result")
plt.axis('off')

plt.tight_layout()
plt.show()
```
