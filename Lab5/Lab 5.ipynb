```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def read_and_convert(filepath):
    img = cv2.imread(filepath)
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

image_path = 'moon.jpg'
```

```python
#Task 1

image_t1 = read_and_convert(image_path)
#7x7 Box filter
smoothed_box = cv2.blur(image_t1, (7,7))

fig1, ax1 = plt.subplots(1, 2, figsize=(10, 5))
ax1[0].imshow(image_t1)
ax1[0].set_title("Original Image (Moon)")
ax1[0].axis("off")

ax1[1].imshow(smoothed_box)
ax1[1].set_title("Smoothed (7x7 Box Filter)")
ax1[1].axis("off")
plt.show()
```

```python
#Task 2

image_t2 = read_and_convert(image_path)

#5x5 and 21x21 Gaussian filters
gaussian_5x5 = cv2.GaussianBlur(image_t2, (5, 5), sigmaX=0)
gaussian_21x21 = cv2.GaussianBlur(image_t2, (21, 21), sigmaX=0)

fig2, ax2 = plt.subplots(1, 3, figsize=(15, 5))
ax2[0].imshow(image_t2)
ax2[0].set_title("Original Image (Moon)")
ax2[0].axis("off")

ax2[1].imshow(gaussian_5x5)
ax2[1].set_title("Gaussian Filter (5x5)")
ax2[1].axis("off")

ax2[2].imshow(gaussian_21x21)
ax2[2].set_title("Gaussian Filter (21x21)")
ax2[2].axis("off")
plt.show()
```

```python
#Task 3

image_t3 = read_and_convert(image_path)

image_float = image_t3.astype(np.float32) / 255.0

laplacian = cv2.Laplacian(image_float, cv2.CV_32F, ksize=3)

sharpened_laplacian = np.clip(image_float - laplacian, 0, 1)

fig3, ax3 = plt.subplots(1, 3, figsize=(15, 5))
ax3[0].imshow(image_t3)
ax3[0].set_title("Original Image (Moon)")
ax3[0].axis("off")

laplacian_display = cv2.normalize(laplacian, None, alpha=0, beta=1, norm_type=cv2.NORM_MINMAX, dtype=cv2.CV_32F)
ax3[1].imshow(laplacian_display)
ax3[1].set_title("Laplacian Filter (Normalized)")
ax3[1].axis("off")

ax3[2].imshow(sharpened_laplacian)
ax3[2].set_title("Sharpened Image")
ax3[2].axis("off")
plt.show()
```
